```python
# Create a DBSCAN model
dbscan = DBSCAN(eps=9, min_samples=5)

# Fit the model and predict cluster labels
X["Cluster"] = dbscan.fit_predict(X)
X["Cluster"]
```



Here is a step-by-step code for explaining DBSCAN:

```python
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)

data_path = RAW_DATA_DIR / "Mall_Customers.csv"
# Read the CSV file into a DataFrame
df = pd.read_csv(data_path)

# Display the first 5 rows
print(df.head().to_markdown(index=False, numalign="left", stralign="left"))

# Print the column names and their data types
print(df.info())

# Extract the features
X = df[["Annual Income (k$)", "Spending Score (1-100)"]]

# Create a scatter plot
chart = (
    alt.Chart(X)
    .mark_point()
    .encode(
        x="Annual Income (k$)",
        y="Spending Score (1-100)",
        tooltip=["Annual Income (k$)", "Spending Score (1-100)"],
    )
    .properties(title="Scatter Plot of Spending Score vs. Annual Income")
    .interactive()
)

# Save the chart
chart.save("spending_score_vs_annual_income_scatter_plot.json")

# Fit a NearestNeighbors model
knn = NearestNeighbors(n_neighbors=5)
knn.fit(X)

# Calculate distances to the 5th nearest neighbor
distances, indices = knn.kneighbors(X)
distances = np.sort(distances[:, 4])

# Create a DataFrame for plotting
df_distances = pd.DataFrame(
    {"Distance": distances, "Point": range(1, len(distances) + 1)}
)

# Create a line chart
chart = (
    alt.Chart(df_distances)
    .mark_line(point=True)
    .encode(x="Point", y="Distance", tooltip=["Point", "Distance"])
    .properties(title="Sorted Distances to the 5th Nearest Neighbor")
    .interactive()
)

# Save the chart
chart.save("sorted_distances_5th_nearest_neighbor.json")

# Create a DBSCAN model
dbscan = DBSCAN(eps=9, min_samples=5)

# Fit the model and predict cluster labels
X["Cluster"] = dbscan.fit_predict(X)

# Count the number of data points in each cluster
cluster_counts = X["Cluster"].value_counts()
print(cluster_counts.to_markdown(numalign="left", stralign="left"))

# Create a scatter plot
chart = (
    alt.Chart(X)
    .mark_point()
    .encode(
        x="Annual Income (k$)",
        y="Spending Score (1-100)",
        color="Cluster:N",
        tooltip=["Annual Income (k$)", "Spending Score (1-100)", "Cluster"],
    )
    .properties(title="DBSCAN Clustering with eps=9 and min_samples=5")
    .interactive()
)

# Save the chart
chart.save("dbscan_clustering_scatter_plot.json")
```
