```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
from sklearn.metrics import silhouette_score
```

```python
# Generate a dataset with 4 blobs
X, y_true = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)
plt.scatter(X[:, 0], X[:, 1], s=50)
plt.title("Generated Data")
plt.show()
```

**2. Implement K-means Clustering**

```python
# Initialize KMeans with 4 clusters
kmeans = KMeans(n_clusters=4)
# Fit the model to the data
kmeans.fit(X)
# Get the cluster labels for each data point
y_kmeans = kmeans.predict(X)
```

**3. Visualize the Clusters**

```python
kmeans.cluster_centers_
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, s=50, cmap="viridis")
# Plot the cluster centers
centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c="black", s=200, alpha=0.5)
plt.title("K-means Clustering (k=4)")
plt.show()
```

**4. Elbow Method**

```python
inertia = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

plt.plot(range(1, 11), inertia, marker="o")
plt.title("Elbow Method for Optimal k")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.show()
```

**5. Silhouette Score**

```python
silhouette_scores = []
for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    score = silhouette_score(X, kmeans.labels_)
    silhouette_scores.append(score)

plt.plot(range(2, 11), silhouette_scores, marker="o")
plt.title("Silhouette Score for Optimal k")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.show()
```

**6. Visualize Cluster Centroid Changes**

```python
# Generate a dataset with 4 blobs
X, y_true = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)

# Initialize KMeans with 4 clusters
kmeans = KMeans(n_clusters=4, random_state=0, max_iter=10)

# Fit the model to the data first
kmeans.fit(X)

# Now you can access cluster_centers_ after fitting the model
centroids = [kmeans.cluster_centers_]

# Get cluster assignments for each data point
labels = kmeans.fit_predict(X)

# Plot the data points and the initial cluster centers
plt.scatter(X[:, 0], X[:, 1], s=50, c=labels, cmap="viridis")  # Use labels for coloring
plt.scatter(
    centroids[0][:, 0],
    centroids[0][:, 1],
    c="red",
    marker="x",
    s=100,
    label="Initial Centroids",
)

# Since we already fit the model above, this loop is not necessary for visualizing initial centroids
# If you want to visualize centroids during iterations, you would need to modify the logic to fit and store centroids within the loop

plt.title("K-means Clustering with Centroid Changes")
plt.legend()
plt.show()
```

This code will visualize the change in cluster centroids after each iteration, providing insights into the convergence of the K-means algorithm.
