**2. Арифметичні операції**

Створіть дві змінні: `var1` зі значенням 15, `var2` зі значенням 4.

Обчисліть та виведіть (кожен пункт оцінюється в 0.5 бала):
- Суму
- Різницю
- Добуток
- Ділення
- Цілочисельне ділення
- Залишок від ділення
- Піднесення до степеня


```python
var1 = 15
var2 = 4

sum = var1 + var2
diff = var1 - var2
product  = var1 * var2
division = var1 / var2
integer_division = var1 // var2
remainder = var1 % var2
power = var1 ** var2

print(sum)
print(diff)
print(product)
print(division)
print(integer_division)
print(remainder)
print(power)
```
