# Thread and Process

```python
import time
import threading

def fetch_user(user_id):
    print(f"Start fetching user {user_id}")
    time.sleep(2)
    print(f"Done fetching user {user_id}")

threads = [
    threading.Thread(target=fetch_user, args=(1,)),
    threading.Thread(target=fetch_user, args=(2,)),
    threading.Thread(target=fetch_user, args=(3,)),
]

for thread in threads:
    thread.start()

for thread in threads:
    thread.join()
```

```python
from concurrent.futures import ThreadPoolExecutor
import time

def fetch_user(user_id):
    print(f"Fetching user {user_id}")
    time.sleep(2)
    return f"User {user_id}"

user_ids = [1, 2, 3, 4, 5]

with ThreadPoolExecutor(max_workers=5) as executor:
    results = executor.map(fetch_user, user_ids)

for result in results:
    print(result)
```

```python
from concurrent.futures import ProcessPoolExecutor

def calculate_sum(limit):
    total = 0

    for number in range(limit):
        total += number

    return total

if __name__ == "__main__":
    limits = [50_000_000, 50_000_000, 50_000_000]

    with ProcessPoolExecutor(max_workers=3) as executor:
        results = executor.map(calculate_sum, limits)

    for result in results:
        print(result)
```

```python
from multiprocessing import Process

def heavy_calculation(name):
    total = 0

    for number in range(50_000_000):
        total += number

    print(f"{name}: {total}")

if __name__ == "__main__":
    process1 = Process(target=heavy_calculation, args=("Process 1",))
    process2 = Process(target=heavy_calculation, args=("Process 2",))

    process1.start()
    process2.start()

    process1.join()
    process2.join()
```
