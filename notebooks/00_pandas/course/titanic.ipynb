# Інтерактивні графіки

## Інтерактивний scatter plot

```python
import plotly.express as px
```

```python
df['family_size'] = df['sibsp'] + df['parch'] + 1
```

```python



# Інтерактивний scatter plot
fig = px.scatter(
    df,
    x='age',
    y='fare',
    color='survived',
    size='family_size',
    hover_data=['name', 'pclass'],
    title='Вік vs Вартість квитка',
    labels={
        'survived': 'Вижив',
        'age': 'Вік',
        'fare': 'Вартість',
    },
    color_continuous_scale='RdYlGn'
)

fig.show()
```

##  Box plot з Plotly

```python
# Box plot з Plotly
fig = px.box(
    df,
    x='pclass',
    y='age',
    color='sex',
    title='Розподіл віку за класом та статтю',
    labels={
        'pclass': 'Клас',
        'age': 'Вік'
    }
)

fig.show()
```

```python
# Sunburst — ієрархічна візуалізація
df_hierarchy = df.groupby(['pclass', 'sex', 'survived']).size().reset_index(name='count')

fig = px.sunburst(
    df_hierarchy,
    path=['pclass', 'sex', 'survived'],
    values='count',
    title='Ієрархія виживання: Клас → Стать → Результат'
)

fig.show()
```

## Heatmap кореляції

```python
# Імпорти
import numpy as np


# Heatmap кореляції
numeric = df.select_dtypes(include=[np.number])
correlation = numeric.corr()

fig = px.imshow(
    correlation,
    text_auto='.2f',
    aspect='auto',
    title='Кореляційна матриця',
    color_continuous_scale='RdBu_r',
    zmin=-1,
    zmax=1
)

fig.show()
```

## 3D scatter plot

```python
# 3D scatter plot
fig = px.scatter_3d(
    df,
    # df.dropna(subset=['age']),
    x='age',
    y='fare',
    z='pclass',
    color='survived',
    hover_data=['name'],
    title='3D візуалізація: Вік, Вартість, Клас',
    labels={
        'survived': 'Вижив'
    }
)

fig.show()
```
