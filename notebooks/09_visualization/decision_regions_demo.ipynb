# Візуалізація decision boundary

Перевіримо перевикористовувану функцію `plot_decision_regions` на двовимірному синтетичному наборі даних.

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler

from ml_homework.visualization import plot_decision_regions
```

## Дані та модель

`random_state` робить приклад відтворюваним. Поліноміальні ознаки дозволяють логістичній регресії побудувати нелінійну межу.

```python
X, y = make_moons(n_samples=240, noise=0.22, random_state=42)

classifier = make_pipeline(
    PolynomialFeatures(degree=3, include_bias=False),
    StandardScaler(),
    LogisticRegression(random_state=42),
)
classifier.fit(X, y)
```

## Decision regions

Світлий фон показує прогноз моделі в кожній частині простору, а насичені точки — фактичні класи спостережень.

```python
figure, axis = plot_decision_regions(
    classifier,
    X,
    y,
    title="Нелінійна межа рішень",
    xlabel="Ознака 1",
    ylabel="Ознака 2",
    class_names={0: "Клас 0", 1: "Клас 1"},
)
plt.show()
```

## Висновок

Модель відтворює вигнуту структуру даних. Точки іншого кольору всередині області класу допомагають одразу побачити помилки класифікації.
