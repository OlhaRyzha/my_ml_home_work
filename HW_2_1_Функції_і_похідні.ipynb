1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import sympy as sp
import matplotlib.pyplot as plt
import numpy as np
```

```python
a = 10
b = 5
h = 0.0001
```

```python
def f(x):
  return a * x - b
```

```python
def derivative(f, x):
    return (f(x + h) - f(x)) / h
```

```python
xs = [x for x in range(0, 11)]
xs
```

```python
derivative_values = [derivative(f, x) for x in xs]
derivative_values
```

```python
f_values = list(map(f, xs))
print(f_values)
```

```python
plt.figure(figsize=(10, 5))
plt.plot(xs, f_values, label="f(x) = 10x - 5", color='red')
plt.plot(xs, derivative_values, label="f'(x) = 10", color='green')

plt.xlabel("Години навчання")
plt.ylabel("Оцінка / Похідна")
plt.title("Функція та її похідна")
plt.legend()
plt.grid(True)

plt.show()
```

Похідна функції f(x) = 10x - 5 дорівнює 10, тобто кожна додаткова година навчання збільшує оцінку на 10 балів. Оскільки похідна стала, то швидкість зміни оцінки не залежить від кількості годин навчання.

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
x = sp.symbols('x')

f_sin = sp.sin(x)
f_log = sp.log(x)


derivative_f_sin = sp.diff(f_sin, x)
derivative_f_log = sp.diff(f_log, x)

print("Похідна функції sin(x):", derivative_f_sin)
print("Похідна функції ln(x):", derivative_f_log)

f_sin_np = sp.lambdify(x, f_sin, 'numpy')
derivative_f_sin_np = sp.lambdify(x, derivative_f_sin, 'numpy')

f_log_np = sp.lambdify(x, f_log, 'numpy')
derivative_f_log_np = sp.lambdify(x, derivative_f_log, 'numpy')


x_sin = np.linspace(0, 10, 100)
x_log = np.linspace(0.1, 10, 100)

plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.plot(x_sin, f_sin_np(x_sin), label='f(x) = sin(x)')
plt.plot(x_sin, derivative_f_sin_np(x_sin), label="f'(x) = cos(x)")
plt.title("Функція sin(x) та її похідна")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(x_log, f_log_np(x_log), label='f(x) = ln(x)')
plt.plot(x_log, derivative_f_log_np(x_log), label="f'(x) = 1/x")
plt.title("Функція ln(x) та її похідна")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
x = sp.symbols('x')

f = x ** 2 - 5 * x + 1
derivative_f = sp.diff(f, x)

print("Функція:", f)
print("Похідна:", derivative_f)

x_values = np.linspace(0, 100, 100)

f_np = sp.lambdify(x, f, 'numpy')
derivative_f_np = sp.lambdify(x, derivative_f, 'numpy')

f_values = f_np(x_values)
derivative_f_values = derivative_f_np(x_values)

plt.plot(x_values, f_values, label='f(x)')
plt.plot(x_values, derivative_f_values, label="f'(x)")
plt.legend()
plt.grid()
plt.show()
```
